```python
import pandas as pd
import os

directory           = 'data'
classroom_file      = 'classrooms.csv'
instructor_file     = 'instructors.csv'
course_file         = 'courses.csv'
student_file        = 'students.csv'
timeslot_file       = 'timeslots.csv'
```

```python
classroom_df = pd.read_csv(os.path.join(directory, classroom_file))
course_df = pd.read_csv(os.path.join(directory, course_file))
instructor_df = pd.read_csv(os.path.join(directory, instructor_file))
student_df = pd.read_csv(os.path.join(directory, student_file))
```

```python
timeslot_df = pd.read_csv(os.path.join(directory, timeslot_file))
display(timeslot_df.head())
```

```python
# Analyze classroom table
#------------------------

display(classroom_df.head())
classroom_check_null = classroom_df.isnull().sum()
if any(classroom_check_null > 0):
    print(classroom_check_null)

n_building = classroom_df.building_name.nunique()
building_room = (
    classroom_df
    .groupby('building_name', as_index=False)
    .agg(
        number_of_rooms = ('classroom_id', 'nunique'), 
        room_type = ('room_type', lambda x: x.unique().tolist()), 
        capacity = ('capacity', lambda x: x.unique().tolist())
    )
)
print(f"Number of buildings: {n_building}")
print(f"Room categories: {classroom_df.room_type.unique().tolist()}")
print(f"Room capacities: {sorted(classroom_df.capacity.unique().tolist())}")
display(building_room)
```

```python
# Analyze course table
#---------------------

display(course_df.head())
course_check_null = course_df.isnull().sum()
if any(course_check_null > 0):
    print(course_check_null)

n_course = course_df.course_name.nunique()
department_course = (
    course_df
    .groupby('department', as_index=False)
    .agg(
        number_of_courses = ('course_name', 'nunique'), 
        credits = ('credits', lambda x: x.unique().tolist()), 
    )
)
print(f"Number of courses: {n_course}")
print(f"Number of departments: {course_df.department.nunique()}")
print(f"Credit categories: {sorted(course_df.credits.unique().tolist())}")
display(department_course)
```

```python
# Analyze instructor table
#-------------------------

display(instructor_df.head())
instructor_check_null = instructor_df.isnull().sum()
if any(instructor_check_null > 0):
    print(instructor_check_null)

n_instructor = instructor_df.instructor_id.nunique()
department_instructor = (
    instructor_df
    .groupby('department', as_index=False)
    .agg(
        number_of_instructors = ('instructor_id', 'nunique'), 
    )
)
print(f"Number of instructors: {n_instructor}")
display(department_instructor)
```

```python
# Analyze students
#-----------------

display(student_df.head())
student_check_null = student_df.isnull().sum()
if any(student_check_null > 0):
    print(student_check_null)

n_student = student_df.student_id.nunique()
n_program = student_df.program_name.nunique()

print(f"Number of students: {n_student}")
print(f"Number of programs: {n_program}")
```
